# Analysis of HT-PAMDA sample data with TAMIPAMI

## Explaination of HT-PAMDA sample data structure

HT-PAMDA provided truncated fastq sequencing files with their code repository.  the daa in `fastq/example_PAM_library` contains the control library.   the data in `fastq/example_PAMDA_data` containd three timepoints of experimental cleavage data.

### Asymetric Illumina 2x75 aired-end formatting 

The  sequencing data i nmade up of  65 nt forawrd reads and 10 nt reverse reads.  Beacuse hese reads do not over lap thera must be processed as single-end data.



### Demultiplexing

The data appear to have inline barcodes that indicate the sample type. The barcodes mapping to sample type are in the file `barcode _csv`

### Replication

Each control or timpoint file has 4 files representing different sequneicng lanes.  I do nok know if there are bioloical or technical replicates.  I am treating them as technical replicates for this reanalysis.
